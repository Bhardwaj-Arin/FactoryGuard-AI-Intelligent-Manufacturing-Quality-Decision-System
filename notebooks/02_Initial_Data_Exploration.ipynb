# Numerical vs Categorical Features

## Objective

Identify the different types of features present in the dataset so that appropriate preprocessing techniques can be applied in later phases of the project.

---

## Feature Categories

The Manufacturing Quality Decisions dataset contains a combination of identifier, datetime, categorical, numerical, and target variables. Understanding these feature types helps determine how each variable should be handled during preprocessing and model development.

### Identifier Feature

The dataset contains one identifier column:

- **event_id**

This column uniquely identifies each manufacturing event. Since it does not provide predictive information, it will be used only for record identification and traceability and will be excluded from model training.

---

### Datetime Feature

The dataset contains one datetime-related feature:

- **event_ts**

This feature records the timestamp of each manufacturing event. During preprocessing, it will be converted to the datetime format and may be used to derive additional time-based features such as month, day, weekday, or production period.

---

### Categorical Features

The dataset contains several categorical variables describing different aspects of the manufacturing environment, including:

- Plant
- Production Line
- Shift
- Machine ID
- Operator ID
- Material Grade
- Inspection Method
- Defect Type

These variables will require appropriate encoding techniques before they can be used for machine learning models.

---

### Numerical Features

The numerical variables describe different operational aspects of the manufacturing process, including:

- Machine age
- Temperature
- Humidity
- Process speed
- Defect severity
- Rework time
- Total cycle time
- Energy consumption
- Production cost

These variables will be used for statistical analysis, feature engineering, visualization, and predictive modeling.

---

### Target Variables

The dataset contains multiple quality-related outcome variables such as:

- final_pass
- decision_rework
- scrap
- warranty_claim_90d

For this project, the final target variable will be selected during the preprocessing phase based on the project's business objective of supporting manufacturing quality decisions.

---

## Summary

The dataset contains a rich combination of identifier, datetime, categorical, numerical, and quality outcome variables. This diverse feature set makes it suitable for developing an intelligent manufacturing analytics system capable of supporting data-driven quality decision-making.
